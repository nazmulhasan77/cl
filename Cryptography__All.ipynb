## 1. Caesar Cipher

```python
# IMPORT
# No external library required

def encrypt(text,shift=3):
    result=""
    for ch in text:
        if ch.isalpha():
            base=ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch)-base+shift)%26+base)
        else:
            result += ch
    return result

def decrypt(text):
    return encrypt(text,-3)

msg=input("Plain text: ")
c=encrypt(msg)
print("Cipher:",c)
print("Original:",decrypt(c))
```

## 2. Polygram Substitution Cipher

```python
# IMPORT
# No external library required

def encrypt(text):
    text=text.replace(" ","").upper()
    cipher=""
    for i in range(0,len(text),3):
        cipher+=text[i:i+3][::-1]
    return cipher

def decrypt(cipher):
    return encrypt(cipher)

msg=input("Plain text: ")
c=encrypt(msg)
print("Cipher:",c)
print("Original:",decrypt(c))
```

## 3. Transposition Cipher

```python
# IMPORT
# No external library required

def encrypt(text,width):
    text=text.replace(" ","")
    while len(text)%width:
        text+="X"
    cipher=""
    for col in range(width):
        for row in range(col,len(text),width):
            cipher+=text[row]
    return cipher

def decrypt(cipher,width):
    rows=len(cipher)//width
    table=[""]*rows
    k=0
    for col in range(width):
        for row in range(rows):
            table[row]+=cipher[k]
            k+=1
    return "".join(table).replace("X","")

msg=input("Message: ")
w=int(input("Width: "))
c=encrypt(msg,w)
print(c)
print(decrypt(c,w))
```

## 4. Double Transposition Cipher

```python
# IMPORT
# Uses previous algorithm idea

def trans(text,w):
    text=text.replace(" ","")
    while len(text)%w:text+="X"
    return ''.join(text[i::w] for i in range(w))

msg=input("Message: ")
w1=int(input("First width: "))
w2=int(input("Second width: "))

c=trans(trans(msg,w1),w2)
print("Cipher:",c)
```

## 5. One Time Pad

```python
# IMPORT
import os
import base64

def otp(data,key):
    return bytes([a^b for a,b in zip(data,key)])

msg=input("Message: ").encode()

key=os.urandom(len(msg))
cipher=otp(msg,key)

print("Cipher:",base64.b64encode(cipher))
print("Original:",otp(cipher,key).decode())
```

## 6. Lehmann Primality Test

```python
# IMPORT
import random

def lehmann(p,k=5):
    for i in range(k):
        a=random.randint(2,p-2)
        r=pow(a,(p-1)//2,p)
        if r!=1 and r!=p-1:
            return False
    return True

n=int(input("Number: "))
print("Prime" if lehmann(n) else "Not Prime")
```

## 7. Miller Rabin Test

```python
# IMPORT
import random

def miller(n,k=5):
    if n<2:return False
    d=n-1
    r=0
    while d%2==0:
        d//=2
        r+=1

    for _ in range(k):
        a=random.randint(2,n-2)
        x=pow(a,d,n)
        if x==1 or x==n-1:
            continue
        return False
    return True

n=int(input("Number: "))
print("Prime" if miller(n) else "Not Prime")
```

```python
from Crypto.PublicKey import RSA
from Crypto.Signature import pkcs1_15
from Crypto.Hash import SHA1
import zlib


# ---- key generation ----
key = RSA.generate(1024)
private_key, public_key = key, key.publickey()


print("N : ", key.n)
print("E : ", key.e)
print("D : ", key.d)


# ----------- Sender Side -----------
# Step 1: Original message
message = "This is a confidential message"

# Step 2: SHA-1 hash
hash_obj = SHA1.new(message.encode())

# Step 3: Encrypt hash with private key
signature = pkcs1_15.new(private_key).sign(hash_obj)

# print( signature)

# Step 4: Concatenate signature + message
packet = signature + message.encode()

# Step 5: Compress
compressed_packet = zlib.compress(packet)

print("Message is sent.")
# print send message and hash
print("Original Message:", message)
print("SHA-1 Hash:", hash_obj.hexdigest())


# ----------- Receiver Side -----------
# Step 6: Decompress
decompressed_packet = zlib.decompress(compressed_packet)

# Step 7: Separate signature and message
received_signature = decompressed_packet[:128]
received_message = decompressed_packet[128:].decode()

# print(f"Received Signature: {received_signature}")

# Step 8: Generate new hash
new_hash = SHA1.new(received_message.encode())

# Step 9: Verify using sender public key
try:
    pkcs1_15.new(public_key).verify(new_hash, received_signature)

    print("\nMessage is AUTHENTIC")
    print("Message:", received_message)
    # print received message and hash
    print("SHA-1 Hash of Received Message:", new_hash.hexdigest())

except (ValueError, TypeError):

    print("Authentication Failed")
    print("Message is NOT authentic")
```

# Confidentiality

```python
# IMPORT
from Crypto.PublicKey import RSA
from Crypto.Cipher import AES, PKCS1_OAEP
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad
import zlib


# -------- Key Generation --------

# Receiver RSA Key
receiver_key = RSA.generate(1024)

receiver_private = receiver_key
receiver_public = receiver_key.publickey()


print("RSA Public Key:")
print(receiver_public.n)
print(receiver_public.e)



# ----------- Sender Side -----------

# Step 1: Original Message

message = "This is a confidential message"


# Step 2: Compress Message

compressed = zlib.compress(message.encode())


# Step 3: Generate AES Session Key

session_key = get_random_bytes(16)



# Step 4: AES Encryption

aes = AES.new(session_key, AES.MODE_ECB)

encrypted_message = aes.encrypt(
    pad(compressed, AES.block_size)
)



# Step 5: Encrypt AES key using Receiver RSA Public Key

rsa = PKCS1_OAEP.new(receiver_public)

encrypted_key = rsa.encrypt(session_key)



print("\nMessage Sent")
print("Encrypted Message:", encrypted_message)
print("Encrypted Session Key:", encrypted_key)



# ----------- Receiver Side -----------


# Step 6: Recover AES Key using RSA Private Key

rsa = PKCS1_OAEP.new(receiver_private)

decrypted_key = rsa.decrypt(encrypted_key)



# Step 7: AES Decryption

aes = AES.new(decrypted_key, AES.MODE_ECB)


decrypted_data = unpad(
    aes.decrypt(encrypted_message),
    AES.block_size
)



# Step 8: Decompress

original_message = zlib.decompress(
    decrypted_data
).decode()



print("\nMessage Received")
print("Original Message:", original_message)
```

PGP Authentication + Confidentiality (SHA-256 + AES)

```python
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
import hashlib


# -----------------------------
# Original Message
# -----------------------------

message = b"Hello PGP SHA and AES"


# -----------------------------
# Authentication (SHA-256)
# -----------------------------

hash_value = hashlib.sha256(message).digest()

print("SHA-256 Hash:")
print(hash_value.hex())



# -----------------------------
# Confidentiality (AES Encryption)
# -----------------------------

# AES Key (16 byte)
key = get_random_bytes(16)

cipher = AES.new(key, AES.MODE_EAX)


# Encrypt message
ciphertext, tag = cipher.encrypt_and_digest(message)


print("\nEncrypted Data:")
print(ciphertext)



# -----------------------------
# Receiver Side
# -----------------------------

# AES Decryption

decrypt_cipher = AES.new(
    key,
    AES.MODE_EAX,
    cipher.nonce
)

decrypted_message = decrypt_cipher.decrypt(ciphertext)


print("\nDecrypted Message:")
print(decrypted_message.decode())



# -----------------------------
# Authentication Check
# -----------------------------

received_hash = hashlib.sha256(
    decrypted_message
).digest()


if hash_value == received_hash:
    print("\nAuthentication Successful")
else:
    print("\nAuthentication Failed")
```

1. PGP Authentication (Digital Signature)

```python
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import hashes


# Generate RSA keys
private_key = rsa.generate_private_key(
    public_exponent=65537,
    key_size=2048
)

public_key = private_key.public_key()


# Message
message = b"Hello PGP Authentication"


# Create Digital Signature
signature = private_key.sign(
    message,
    padding.PSS(
        mgf=padding.MGF1(hashes.SHA256()),
        salt_length=padding.PSS.MAX_LENGTH
    ),
    hashes.SHA256()
)

print("Signature Generated")


# Verify Signature
try:
    public_key.verify(
        signature,
        message,
        padding.PSS(
            mgf=padding.MGF1(hashes.SHA256()),
            salt_length=padding.PSS.MAX_LENGTH
        ),
        hashes.SHA256()
    )

    print("Authentication Successful")

except:
    print("Authentication Failed")
```

4. PGP Authentication + Confidentiality

```python
from cryptography.fernet import Fernet
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import hashes


# ------------------
# Confidentiality
# ------------------

key = Fernet.generate_key()

cipher = Fernet(key)


message = b"PGP Authentication and Confidentiality"


# Encrypt Message
encrypted_message = cipher.encrypt(message)


print("Encrypted Message:")
print(encrypted_message)



# ------------------
# Authentication
# ------------------

# Generate RSA Keys

private_key = rsa.generate_private_key(
    public_exponent=65537,
    key_size=2048
)


public_key = private_key.public_key()



# Generate Signature

signature = private_key.sign(
    encrypted_message,
    padding.PSS(
        mgf=padding.MGF1(hashes.SHA256()),
        salt_length=padding.PSS.MAX_LENGTH
    ),
    hashes.SHA256()
)


print("\nDigital Signature Created")



# Verify Signature

try:

    public_key.verify(
        signature,
        encrypted_message,
        padding.PSS(
            mgf=padding.MGF1(hashes.SHA256()),
            salt_length=padding.PSS.MAX_LENGTH
        ),
        hashes.SHA256()
    )

    print("Authentication Successful")


except:

    print("Authentication Failed")



# Decrypt Data

decrypted_message = cipher.decrypt(encrypted_message)

print("\nOriginal Message:")
print(decrypted_message.decode())
```
